```python
import numpy as np
import pylab as plt
from scipy.stats import norm
```

```python
mean = 1
std = 0.2
N = 8

sample = np.random.normal(loc=mean, scale=std, size=N)
print(sample)

lin = np.linspace(0, 2, 10000)

L = []

for i in sample:
    L.append(norm.pdf(lin ,loc=i ,scale=std))

#print(L)

likelihood = np.prod(L ,axis=0)

for i in L:
    plt.plot(lin, i)
    
plt.plot(lin, likelihood)
```

```python
print(lin[np.argsort(likelihood)[-1]])
print(np.mean(sample))
```

```python
max = np.argsort(likelihood)[-1]
print(max)
```

```python

std_mean = 1/np.sqrt(-np.diff(np.log(likelihood), n=2)/(lin[1] - lin[0])**2)[max]
print(std_mean)
```
