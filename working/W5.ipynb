Likelyhood, Fisher information matrix

PART 1

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
```

```python
mu = 1
std_of_std = 0
np.random.seed(0)               #19.8 normalization works with std_of_std = 0 (part 1 and 2)

N = 8
sigma = np.random.normal(0.2, std_of_std, N)

sample = np.concatenate([np.random.normal(mu,x, 1) for x in sigma])

lin = np.linspace(0, 2, 10000)

L = [norm.pdf(lin, loc=x, scale=y) for x,y in zip(sample,sigma)]

likelihood = np.prod(L ,axis=0)

for i in L:
    plt.plot(lin, i)
    
plt.plot(lin, likelihood)
```

```python
print(lin[np.argsort(likelihood)[-1]])
MLest = np.average(sample, weights=1/sigma**2)
print(MLest)
```

PART 2

```python
max = np.argsort(likelihood)[-1]
```

```python
sigma_mu = 1/np.sqrt(-np.diff(np.log(likelihood), n=2)/(lin[1] - lin[0])**2)[max]
print(sigma_mu)
MLest_std = np.sum(sigma**-2)**-0.5
print(MLest_std)
```

```python
plt.plot(lin, likelihood)

new_gaussian = norm.pdf(lin ,loc=MLest, scale=MLest_std)
plt.plot(lin, 19.8*new_gaussian)
```
