Markov Chain

```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
day = []
day.append(1)       #0 clear, 1 cloudy

choice = [0, 1]     #clear/cloudy
for i in tqdm(range(1000000)):
    if day[i] == 0:
        day.append(np.random.choice(choice, p=[0.5, 0.5]))
    if day[i] == 1:
        day.append(np.random.choice(choice, p=[0.1, 0.9]))

day = np.array(day)
```

```python
weather = np.cumsum(day)/(np.arange(day.shape[0])+1)
plt.plot(weather)
plt.xscale("log")   #usefull do determine how long it takes for the traceplot to converge
```

```python
plt.hist(weather, bins=316, density=True);
print(np.mean(weather))
print(np.std(weather))
```

```python
#the traceplot converges often at 4*10**4
burn = 40000
weather = weather[burn:]
plt.hist(weather, bins=316, density=True);
print(np.mean(weather))
print(np.std(weather))
```
