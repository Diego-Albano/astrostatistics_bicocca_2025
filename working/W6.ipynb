Gaussian Mixture Model, Akaike information criterion

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
```

```python
plt.rcParams['text.usetex'] = False             #non so perché ma matplotlib vuole usare latex
```

```python
x = np.load("formationchannels.npy")
```

```python
plt.hist(x, bins=50);
```

```python
GaussianMixture().fit(x)
```

```python
Ns = np.arange(1, 10)

models=[GaussianMixture(N).fit(x) for N in Ns]
```

```python
#akaike information criterion
AIC = [m.aic(x) for m in models]
```

```python
plt.plot(Ns, AIC)
```

```python
plt.hist(x, bins=50, density=True);

lin = np.linspace(0, 60, 1000)
logprob = models[2].score_samples(lin.reshape(-1, 1))
pdf = np.exp(logprob)
plt.plot(lin, pdf)
```
